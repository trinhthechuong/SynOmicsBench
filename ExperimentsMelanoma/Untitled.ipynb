```python
import pandas as pd
import numpy as np
omics_data = pd.read_csv("MelanomaData/TranscriptomicsData.csv", index_col = 0)
omics_data.head()
```

```python
expression_matrix = omics_data.T
sum_expression = expression_matrix.sum(axis = 1)
zero_genes = sum_expression[sum_expression == 0].keys().tolist()

variances_expression = expression_matrix.var(axis = 1)
near_zero_var = variances_expression[variances_expression <= 0.0005].keys().tolist()

remove_genes = set(zero_genes) | set(near_zero_var)
```
