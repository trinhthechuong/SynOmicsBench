```python
import pandas as pd
import numpy as np
from ctgan import CTGAN
import random
import torch

seed = 42
random.seed(seed)
np.random.seed(seed)
torch.manual_seed(seed)
model = CTGAN.load('CTGAN_model.pkl')
# original_data = pd.read_csv("../Notebook/Integration_pipeline_JOBIM/integrated_data_jobim.csv", index_col = 0)
```

```python
import matplotlib.pyplot as plt

generator_losses = model.loss_values['Generator Loss'].values.tolist()
discriminator_losses = model.loss_values['Discriminator Loss'].values.tolist()

# Plotting
plt.figure(figsize=(10, 6))
plt.plot(generator_losses, label='Generator Loss', color='blue')
plt.plot(discriminator_losses, label='Discriminator Loss', color='red')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('CTGAN Generator vs. Discriminator Loss')
plt.legend()
plt.grid(True)
plt.tight_layout()
# plt.savefig("CTGAN_Loss_200epochs_2307.png", dpi=300,bbox_inches='tight')
plt.show()
```
