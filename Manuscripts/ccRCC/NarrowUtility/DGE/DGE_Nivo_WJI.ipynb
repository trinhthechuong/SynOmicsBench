```python
import pandas as pd
import numpy as np
import os

seeds = [42,0,1,2,3]
seed = 42

dges_results = {}
original_dge = pd.read_csv('DGE_Nivo/DGE_Origin.csv')
dges_results['Origin'] = original_dge
synthetic_datasets = [f"avatarsk5_{seed}",f"avatarsk10_{seed}",f"ctgan_{seed}",f"gaussiancopula_{seed}",
                      f"synthpop_{seed}",f"tvae_{seed}"]
or_dge = dges_results['Origin'].set_index('Gene')['Q_value'].to_dict()
synth_w = {}
for synthetic_dataset in synthetic_datasets:
    try:
        dge_syn_result = pd.read_csv(f'../DGE/DGE_Nivo/Seed_{seed}/DGE_{synthetic_dataset}.csv')
        dges_results[synthetic_dataset] = dge_syn_result
        synth_w[synthetic_dataset] = dge_syn_result.set_index('Gene')['Q_value'].to_dict()
    except Exception as err:
        print(f"{synthetic_dataset}: {err}")
```

```python
or_dge = dges_results['Origin'].set_index('Gene')['Q_value'].to_dict()
syn_dge = dges_results['ctgan_42'].set_index('Gene')['Q_value'].to_dict()
```

```python
import logging
from WeightedJaccardIndex import compute_weighted_jaccard_indices

logging.basicConfig(level=logging.INFO)

df = compute_weighted_jaccard_indices(
    original_pathways=or_dge,
    synth_pathways_dict=synth_w,
    n_permutations=1000,
    seed=42,
    verbose=2,  # 0/1/2
)
```

```python
df.to_csv(f"DGE_Nivo/Seed_{seed}/JaccardIndex_{seed}_NivoBenefit.csv", index=False)
```
