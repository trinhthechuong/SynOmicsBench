```python
import pandas as pd
import numpy as np
import sys
parent_path = '/mnt/digphat/syntheticDataBenchmark/Chuong_pipeline'
sys.path.append(parent_path)
from SynOmics.metrics.fidelity.PairwiseSimilarity import PairwiseSimilarity
from SynOmics.processing.metadata import MetaData
from SynOmics.processing.postprocessing import post_masking
from BroadUtility import plot_violin

or_data = pd.read_csv("../Data/original_data.csv", index_col = 0)
clinical_or_data = or_data.iloc[:,0:28]
masked_clinical_or_data = post_masking(clinical_or_data)

ordinal_cat_columns =  ["Patient_Smoking_Status", "Line_of_Therapy"]
metadata = MetaData.get_metadata(data = masked_clinical_or_data, 
                                 threshold_unique_values = 10, 
                                    ordinal_features = ordinal_cat_columns)
seeds = [42,0,1,2,3]
```

```python
def save_results(path, result):
    result_arr = np.asarray(result)
    np.save(path, result_arr)
```

```python
import matplotlib.pyplot as plt
results_all = {}
for seed in seeds:
    print(f"---Seed {seed}---")
    syn_datas = [f"avatarsk5_{seed}",f"avatarsk10_{seed}",f"ctgan_{seed}",f"gaussiancopula_{seed}",f"synthpop_{seed}",f"tvae_{seed}"]
    results_seed = {}
    visulize_results_dict = {}
    visulize_results_high = {}
    for syn_data in syn_datas:
        print(f"---Method {syn_data}---")
        syn_df = pd.read_csv(f"../Data/{syn_data}.csv", index_col = 0)
        clinical_syn_df = syn_df.iloc[:,0:28]
        masked_syn_clinical_df = post_masking(clinical_syn_df)
        pw_computer = PairwiseSimilarity(
        original_data = masked_clinical_or_data,
        synthetic_data = masked_syn_clinical_df,
        metadata = metadata,
        output_dir = f"PairwiseClinical/PairwiseClinical_{seed}",
        verbose = True,
        save = True,
        name = syn_data
    )
        result = pw_computer.get_pairwise_scores(method = "spearman")
        results_seed[syn_data] =result
        visulize_results_dict[syn_data] = result["PairwiseScore"]

        #Filtering Correlations 
        or_correlations = result["OriginalCorrelation"]
        filtered_indices =  np.where(np.absolute(or_correlations) > 0.5)[0]
        pairwise_scores = result["PairwiseScore"]
        filtered_pairwise_scores = pairwise_scores[filtered_indices]
        visulize_results_high[syn_data] = filtered_pairwise_scores

        path_score = f"PairwiseClinical/PairwiseClinical_{seed}/bivariate_{syn_data}_all.npy"
        save_results(path_score, pairwise_scores)

        path_score_filtered = f"PairwiseClinical/PairwiseClinical_{seed}/bivariate_{syn_data}_filtered.npy"
        save_results(path_score_filtered, filtered_pairwise_scores)

        
    title = f'Distribution of pairwise similarity scores (Clinical Features)'
    fig, ax = plot_violin(
    visulize_results_dict,
    title=title,
    annotate=True,
    figsize=(10, 5),
)
    fig.savefig(f"PairwiseClinical/PairwiseClinical_{seed}/Benchmark_PairwiseClinical_{seed}.png", dpi=300, bbox_inches='tight')
    plt.show()

    title = f'Distribution of pairwise similarity scores (Clinical Features)'
    fig, ax = plot_violin(
    visulize_results_high,
    title=title,
    annotate=True,
    figsize=(10, 5),
)
    # print("Pairwise p-values used for heatmap (DataFrame):")
    # print(pvals.round(4))
    fig.savefig(f"PairwiseClinical/PairwiseClinical_{seed}/Benchmark_PairwiseClinical_Filtered_{seed}.png", dpi=300, bbox_inches='tight')
    plt.show()
    # results_all[seed] = results_seed    
```
