## Reproduce signature tables

```python
import pandas as pd
from scipy.stats import zscore
import warnings

def signature_score(expression_data, signatures):
    """
    expression_data: DataFrame (Hàng là Patient_ID, Cột là Gene)
    signatures: Dictionary chứa tên signature và danh sách gene tương ứng
    """
    results = {}
    
    for sig_name, genes in signatures.items():
        valid_genes = [g for g in genes if g in expression_data.columns]
        
        if len(valid_genes) == 0:
            print(f"Warning: Không tìm thấy gene nào của signature '{sig_name}' trong dữ liệu.")
            continue
            
        sig_exp = expression_data[valid_genes]
        
        avg_expression = sig_exp.mean(axis=1, skipna=True)
        
        results[sig_name] = zscore(avg_expression)
        
    output_df = pd.DataFrame(results, index=expression_data.index)
    return output_df

def tpm_to_log2tpm(tpm,
                   pseudo_count: float = 1.0,
                   fillna_with_zero: bool = True):

    if isinstance(tpm, pd.DataFrame):
        mat = tpm.copy()
        if fillna_with_zero:
            mat = mat.fillna(0.0)
        return np.log2(mat + pseudo_count)
    else:
        arr = np.array(tpm, dtype=float, copy=True)
        if fillna_with_zero:
            # replace NaN with 0
            arr = np.nan_to_num(arr, nan=0.0)
        return np.log2(arr + pseudo_count)
```

```python

        
seeds = [0,1, 2, 3, 42]  # seeds/replicates to include
method_bases = {
    'Avatars K5': 'avatarsk5',
    'Avatars K10': 'avatarsk10',
    'CTGAN': 'ctgan',
    'Gaussian Copula': 'gaussiancopula',
    'Synthpop': 'synthpop',
    'TVAE': 'tvae'
}
for name, base in method_bases.items():
    dfs = []
    for s in seeds:
        fpath = f'../../Data/{base}_{s}.csv'
        try:
            orginal_data = pd.read_csv(fpath, index_col = 0)
        except Exception as e:
            warnings.warn(f"Could not read {fpath}: {e}")
            continue

        expression_matrix = orginal_data.iloc[:, 28:]
        expression_matrix = expression_matrix.T
        expression_matrix = (expression_matrix
                             .replace([np.inf, -np.inf], 0)
                             .fillna(0)
                             .clip(lower=0)).T

        log2tpm_df = tpm_to_log2tpm(expression_matrix,pseudo_count=1)

        # Cách gọi hàm:
        zi_scores  = signature_score(log2tpm_df, Reference_metagenes_ZI)
        sf_scores = signature_score(log2tpm_df, Reference_metagenes_SF)
        other_scores = signature_score(log2tpm_df, Ref_others)
        dataset = f"{base}_{s}"
        zi_scores.to_csv(f'Signature_analysis/ZI_{dataset}.csv', index = True)
        
        sf_scores.to_csv(f'Signature_analysis/SF_{dataset}.csv', index = True)
        
        other_scores.to_csv(f'Signature_analysis/Other_{dataset}.csv', index = True)

        


        
        
        
```
