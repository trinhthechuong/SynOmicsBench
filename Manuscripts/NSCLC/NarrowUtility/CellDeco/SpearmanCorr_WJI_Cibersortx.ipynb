```python
import pandas as pd
import numpy as np

from Utils import spearman_across_datasets

# DA_results_cibersortx = {}
seeds = [42,0,1,2,3]


for seed in seeds:
    print(f"\n{'='*60}")
    print(f"Seed: {seed}")
    print(f"{'='*60}")
    DA_results_cibersortx = {}
    DA_res = pd.read_csv(f'ResultsDA_cibersortx/Origin.csv')
    DA_results_cibersortx['Origin']=DA_res
    
    datasets = [
        f'avatarsk5_{seed}', 
        f'avatarsk10_{seed}',
        f'ctgan_{seed}',
        f'gaussiancopula_{seed}', 
        f'synthpop_{seed}', 
        f'tvae_{seed}'
    ]
    
    for dataset in datasets:
        DA_res = pd.read_csv(f'ResultsDA_cibersortx/Seed_{seed}/{dataset}.csv')
        DA_results_cibersortx[dataset]=DA_res

    spearman_df, _ = spearman_across_datasets(
    gsea_overall=DA_results_cibersortx,
    origin="Origin",
    term_col="Cell Type",
    score_col="Diff_Mean",
    min_terms=3)

    spearman_df.to_csv(f'ResultsDA_cibersortx/Seed_{seed}/Spearman_{seed}.csv')
```

```python
from WeightedJaccardIndex import compute_weighted_jaccard_indices
seeds = [42,0,1,2,3]
for seed in seeds:
    print(f"\n{'='*60}")
    print(f"Seed: {seed}")
    print(f"{'='*60}")
    DA_results_cibersortx = {}
    
    datasets = [
        f'avatarsk5_{seed}', 
        f'avatarsk10_{seed}',
        f'ctgan_{seed}',
        f'gaussiancopula_{seed}', 
        f'synthpop_{seed}', 
        f'tvae_{seed}'
    ]
    or_DA_res = pd.read_csv(f'ResultsDA_cibersortx/Origin.csv')
    or_w = or_DA_res.set_index('Cell Type')['Q_values'].to_dict()
    q_values_syn = {}
    for dataset in datasets:
        DA_res_syn = pd.read_csv(f'ResultsDA_cibersortx/Seed_{seed}/{dataset}.csv')
        syn_q_value = DA_res_syn.set_index('Cell Type')['Q_values'].to_dict()
        q_values_syn[dataset] = syn_q_value

    df_Jaccard = compute_weighted_jaccard_indices(
        original_pathways=or_w,
        synth_pathways_dict=q_values_syn,
        n_permutations=1000,
        seed=42,
        verbose=0,  # 0/1/2
    )
    # Save df_Jaccard
    df_Jaccard.to_csv(f"ResultsDA_cibersortx/Seed_{seed}/WeightedJaccardIndex_{seed}.csv", index=False)
```
