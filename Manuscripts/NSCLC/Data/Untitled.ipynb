```python
import pandas as pd 

seeds = [0,1,2,3,42]
# seeds = [0]
or_data = pd.read_csv("original_data.csv", index_col = 0)
ordered_cols = or_data.columns.tolist()
for seed in seeds:
    data = pd.read_csv(f"synthpop_{seed}.csv", index_col = 0)
    ordered_data = data[ordered_cols]
    ordered_data.to_csv(f"synthpop_{seed}.csv", index=True)
```
