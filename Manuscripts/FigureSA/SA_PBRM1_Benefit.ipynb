```python
import os
import pandas as pd
import numpy as np
import sys
from visulize_grid import plot_survival_grid
import matplotlib.pyplot as plt
import scienceplots

# --------- User Parameters ---------
seeds = [0, 1, 2, 3, 42]
methods_order = ["Avatars K5", "Avatars K10", "CTGAN", "Gaussian Copula", "Synthpop", "TVAE"]
phenotype = {"PBRM1": ["MUT", "WT"]}
original_data_path = "../ccRCC/Data/original_data.csv"
synthetic_pattern = "../ccRCC/Data/{}_{}.csv"  # e.g. avatarsk5_0.csv

# --------- Main Loop Over All Seeds ---------
for seed in seeds:
    dataset_dict = {}
    # ---- Load original ----
    if not os.path.exists(original_data_path):
        raise FileNotFoundError(f"Missing original: {original_data_path}")
    dataset_dict["Origin"] = pd.read_csv(original_data_path, index_col=0)

    # ---- Load synthetic for this seed ----
    syn_datas = [f"avatarsk5_{seed}", f"avatarsk10_{seed}", f"ctgan_{seed}", f"gaussiancopula_{seed}", f"synthpop_{seed}", f"tvae_{seed}"]
    for i, tool in enumerate(syn_datas):
        syn_file = f"../ccRCC/Data/{tool}.csv"
        if not os.path.exists(syn_file):
            raise FileNotFoundError(f"Missing synthetic file: {syn_file}")
        dataset_dict[methods_order[i]] = pd.read_csv(syn_file, index_col=0)

    # ---- Plot grid ----
    fig, summary = plot_survival_grid(
        datasets_dict=dataset_dict,
        phenotype=phenotype,
        treatment=None,
        time_target="OS",
        event_target="OS_CNSR",
        figsize=(18, 7),
        show_censors=False,
        ci_show=False,
        title_prefix="ccRCC PBRM1"
    )
    print(f"[Seed {seed}]")

    # ---- Setup Output Directory ----
    output_dir = f"Figure8a_PBRM1/Seed_{seed}/"
    os.makedirs(output_dir, exist_ok=True)

    # ---- Save grid plot ----
    fig.savefig(os.path.join(output_dir, "OS_KMCurve_grid.pdf"), dpi=300, bbox_inches="tight")
    fig.savefig(os.path.join(output_dir, "OS_KMCurve_grid.png"), dpi=400, bbox_inches="tight")
```
