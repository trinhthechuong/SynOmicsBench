## **Clean Transcriptomics Data**

```python
su2c_rna_harm = pd.read_csv(source_data_path_rna + 'SU2C-MARK_Harmonized_rnaseqc_tpm_v1.gct',skiprows=2,sep='\t')
su2c_rna_harm.head()
```

```python
#Check descriptions
hugo_ids = su2c_rna_harm['Description'].values.tolist()
su2c_rna_ensemble = su2c_rna_harm.drop(columns = ["Name"])
su2c_rna_ensemble.head()
```

```python
expression_matrix = su2c_rna_ensemble.set_index('Description')
sum_expression = expression_matrix.sum(axis = 1)
zero_genes = sum_expression[sum_expression == 0].keys().tolist()

variances_expression = expression_matrix.var(axis = 1)
near_zero_var = variances_expression[variances_expression <= 0.01].keys().tolist()

remove_genes = set(zero_genes) | set(near_zero_var)

cleaned_expression_matrix = expression_matrix.drop(index = remove_genes)
# cleaned_expression_matrix.to_csv('cleaned_rna_expression.csv', index = True)
```

```python
cleaned_expression_matrix
```

```python
log2tpm_matrix = np.log2(cleaned_expression_matrix + 1)

log2tpm_threshold = 0.5
min_sample_percent = 0.30

# Calculate percentage of samples where each gene passes threshold (axis=1 for row-wise)
percent_expressed = (log2tpm_matrix > log2tpm_threshold).sum(axis=1) / log2tpm_matrix.shape[1]

# Create filter:  keep genes expressed in >= 30% of samples
genes_to_keep = percent_expressed >= min_sample_percent

# Step 3: Apply filter to original TPM matrix
filtered_tpm_matrix = cleaned_expression_matrix[genes_to_keep]

# Print results
print(f"Original genes: {len(cleaned_expression_matrix)}")
print(f"Filtered genes: {len(filtered_tpm_matrix)}")
print(f"Genes removed: {len(cleaned_expression_matrix) - len(filtered_tpm_matrix)}")

# Save filtered TPM matrix
# filtered_tpm_matrix.to_csv('filtered_tpm_matrix.csv')
```

```python
cleaned_expression_matrix_t = filtered_tpm_matrix.T
cleaned_expression_matrix_t.to_csv('cleaned_rna_expression.csv', index = True)
```
