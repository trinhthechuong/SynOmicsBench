## Avatars K5

```python
from pathlib import Path
import pandas as pd
import numpy as np

base_dir = Path("synthetic_blocks_ccRCC_k5_42")

file_paths = list(base_dir.glob("emissions_block_*.csv"))

memory_path = list(base_dir.glob("memory_history_block_*.csv"))

cpu_usages = [pd.read_csv(p)['cpu_utilization_percent'].values[0] for p in file_paths]

durations = [pd.read_csv(p)['duration'].values[0] for p in file_paths]

enery_consumption = [pd.read_csv(p)['energy_consumed'].values[0] for p in file_paths]

peak_mem = [pd.read_csv(p)['mem_usage_mib'].max() for p in memory_path]

print(f"Duration: {np.mean(durations):.0f} s")
print(f"Energy Consumption: {np.mean(enery_consumption):.4f} kWh")
print(f"CPU Mean: {np.mean(cpu_usages):.1f} %")
print(f"Mean Max Peak: {np.mean(peak_mem):.1f} MiB")
print("-" * 40)


    
```
