```python
def get_learned_distributions(model):
    learned = {}
    for col, univariate in zip(model.columns, model.univariates):
        try:
            ud = univariate.to_dict()
        except Exception:
            # Fallback to introspection if to_dict is not available
            ud = {}
        # Try to be flexible with different Univariate.to_dict formats
        dist_name = ud.get("type") or ud.get("distribution") or univariate.__class__.__name__
        # learned parameters could be stored in different keys; try common possibilities
        params = ud.get("parameters") or ud.get("learned_parameters") or {
            k: v for k, v in ud.items() if k not in ("type", "distribution")
        }
        # Ensure params is a dict
        if not isinstance(params, dict):
            params = {"value": params}
        learned[col] = {
            "distribution": dist_name,
            "learned_parameters": params,
        }
    return learned
```

```python
learned_distribution_1 = get_learned_distributions(model_debug1)
learned_distribution_42 = model_42.get_learned_distributions()
```
