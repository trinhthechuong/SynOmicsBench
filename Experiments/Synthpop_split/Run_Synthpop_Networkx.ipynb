```python
from rpy2.robjects import conversion, default_converter
import rpy2.robjects as robjects
from rpy2.robjects.pandas2ri import converter as pandas2ri_converter
import os
r_home = "/opt/R/4.4.1/lib/R"
os.environ["R_HOME"] = r_home
r_terminal = "R441"


robjects.r(".libPaths()")

r_code = """
        suppressPackageStartupMessages(library(synthpop))
        create_synthetic_data <- function(data, seed, discrete_columns, minnumlevels, 
                                          method, proper, m, k, visit_sequence, cont_na, verbose,
                                          predictorMatrix = NULL) {
            # Ensure discrete (categorical) features are factors
            cat_features <- discrete_columns
            cat_features <- trimws(cat_features)
            if (length(cat_features) > 0) {
                for (cn in cat_features) {
                    if (cn %in% names(data)) {
                        data[[cn]] <- as.factor(data[[cn]])
                    }
                }
            }

            # Visit sequence
            if (is.null(visit_sequence)) {
                visit_seq <- colnames(data)
            } else {
                visit_seq <- trimws(visit_sequence)
            }

            # cont.na pass-through
            if (is.null(cont_na)) {
                cont_na <- NULL
            }

            # Predictor matrix handling
            pred_mat <- NULL
            if (!is.null(predictorMatrix)) {
                # Ensure matrix with dimnames; if rownames missing, copy from colnames
                pred_mat <- as.matrix(predictorMatrix)
                if (is.null(rownames(pred_mat)) || any(rownames(pred_mat) != colnames(pred_mat))) {
                    rownames(pred_mat) <- colnames(pred_mat)
                }
            }

            set.seed(seed)
            res <- syn(
                data,
                minnumlevels = minnumlevels,
                method = method,
                proper = proper,
                m = m,
                k = k,
                visit.sequence = visit_seq,
                cont.na = cont_na,
                print.flag = verbose,
                predictor.matrix = pred_mat
            )

            # Extract synthetic data(s)
            synth_data <- res$syn

            # Convert all factor columns to character
            if (m > 1) {
                synth_data <- lapply(synth_data, function(df) {
                    df[] <- lapply(df, function(x) if (is.factor(x)) as.character(x) else x)
                    df
                })
            } else {
                synth_data[] <- lapply(synth_data, function(x) if (is.factor(x)) as.character(x) else x)
            }
            synth_data
        }
        """
robjects.r(r_code)
r_func = robjects.globalenv["create_synthetic_data"]
```
